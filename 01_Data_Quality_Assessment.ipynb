# Data Organization and Feature Grouping

## 1. Overview

As part of the initial data preparation process, the original airline dataset was examined and reorganized into several smaller, logically structured subsets.

The primary objective of this step was to improve data organization, simplify subsequent analysis, and group related variables based on their analytical purpose.

Since the main objective of this project is to investigate flight delays and identify the factors contributing to differences in airline and airport performance, the dataset was divided into subsets representing different aspects of flight operations.

## 2. Dataset Structure

The original dataset initially contained **110 columns**.

After reviewing the available features, the columns were organized into **7 separate DataFrames**, each representing a specific analytical category.

### Created DataFrames

| No. | DataFrame Name | Description |
|---|---|---|
| 1 | `df_Time` | Contains temporal features, including year, quarter, month, day, and flight date. |
| 2 | `df_Airline` | Contains airline identification, aircraft registration, flight numbers, and origin and destination airport information. |
| 3 | `Delay_Time` | Contains scheduled and actual departure and arrival times, along with delay measurements and delay categories. |
| 4 | `df_Performance` | Contains operational performance variables, including taxi times, flight duration, distance, and flight activity. |
| 5 | `df_Cancellation_Diversion` | Contains flight cancellation indicators, cancellation reasons, diversion indicators, and related operational information. |
| 6 | `df_Delay_Causes` | Contains the five primary delay cause variables: carrier, weather, NAS, security, and late aircraft delays. |
| 7 | `df_Diversion` | Contains additional operational information about diverted flights, including diversion airport details and ground times. |
| 8 | `df_Additional_Diversion` | Contains supplementary information about subsequent diversion airports, covering the second through fifth diversion locations. |

## 3. Feature Grouping Process

The grouping process was performed according to the meaning and analytical purpose of each variable.

The main categories included:

- **Temporal Features:** Variables used to investigate daily, monthly, quarterly, and yearly flight patterns.
- **Airline and Route Features:** Variables used to compare airlines, airports, and flight routes.
- **Delay Time Features:** Variables used to measure departure and arrival delays.
- **Flight Performance Features:** Variables used to examine flight duration, distance, and ground operations.
- **Cancellation and Diversion Features:** Variables used to investigate cancelled and diverted flights.
- **Delay Cause Features:** Variables used to identify and compare the primary causes of flight delays.
- **Additional Diversion Features:** Variables used to examine operational details of flights involving multiple diversion locations.

## 4. Data Quality and Column Validation

After creating the subsets, a validation process was performed to ensure that all original dataset columns had been assigned to the appropriate groups.

The validation process identified:

- Initially, 24 columns had not been assigned to any subset.
- These columns were reviewed and distributed among the relevant DataFrames.
- One completely empty column, `Unnamed: 109`, was identified and removed from the original dataset.
- Two duplicated assignments, `Div1WheelsOff` and `Div1LongestGTime`, were identified and corrected to avoid unnecessary repetition across subsets.

### Final Validation Results

| Validation Metric | Result |
|---|---:|
| Original dataset columns after cleanup | 109 |
| Unique columns assigned to subsets | 109 |
| Unassigned columns | 0 |
| Duplicated column assignments | 0 |
| Completely empty columns removed | 1 |

## 5. Conclusion

The initial dataset organization and feature grouping process has been completed successfully.

All 109 remaining columns have been assigned to their corresponding analytical subsets, with no unassigned or duplicated column assignments.

This structured organization provides a clearer foundation for the next stages of the project, including:

### 3. Investigation and Findings

Missing values were investigated by grouping flights based on their cancellation and diversion status.

#### Normal Flights

A total of 594,207 flights were neither cancelled nor diverted.

All six operational performance columns were complete for these flights.

#### Diverted Flights

A total of 2,869 flights were diverted.

- `AirTime` and `ActualElapsedTime`: All 2,869 records were missing.
- `TaxiIn` and `WheelsOn`: 461 records were missing.
- `WheelsOff` and `TaxiOut`: No missing values were found.

#### Cancelled Flights

A total of 11,313 flights were cancelled.

- `AirTime`, `ActualElapsedTime`, `TaxiIn`, and `WheelsOn`: All 11,313 records were missing.
- `WheelsOff` and `TaxiOut`: 11,260 records were missing.

An additional investigation identified 53 cancelled flights with available `TaxiOut` and `WheelsOff` values.

These flights had a special operational status: they departed the gate, took off, and subsequently returned to an airport before being cancelled.

According to the BTS reporting guidelines, the value `9` in `DivAirportLandings` identifies an air return where the flight was ultimately cancelled. The corresponding diverted-airport fields record information about the return landing.

### 4. Columns Without Missing Values

The following columns contained no missing values:

- `CRSElapsedTime`
- `Distance`
- `DistanceGroup`
- `Flights`

### 5. Data Cleaning Decision

No missing values were imputed or removed during this analysis.

The identified missing values are largely associated with operational circumstances, including flight cancellations and diversions.

All records were retained to preserve potentially valuable information about airline operational performance.

### 6. Conclusion

The missing value analysis revealed clear relationships between missing operational performance data and flight status.

Normal flights had complete operational records, while cancelled and diverted flights accounted for the missing values.

The 53 air-return cases also demonstrated why flight status alone is not sufficient to explain missing operational information.

No imputation was performed at this stage.

# Missing Value Analysis – Diversion Data

## 1. Objective

The purpose of this analysis was to investigate missing values in flight diversion records and determine whether they represent data quality issues or are related to actual flight operations.

## 2. Analysis of Diverted Flights

A total of **2,869 diverted flights** were identified:

- 2,408 flights eventually reached their original destinations.
- 461 flights did not reach their original destinations.

The results showed that all diverted flights had information about their first alternate airport and landing time.

However, the availability of departure time from the alternate airport differed:

- All 2,408 flights that reached their original destinations had a recorded departure time from the first alternate airport.
- Among the 461 flights that did not reach their original destinations, only 9 had this information recorded.

This suggests that missing departure information is related to the operational status of the flights and should not automatically be treated as a data quality issue.

## 3. Analysis of Flights with Two Alternate Airports

A total of **12 flights** had two alternate airport landings.

The investigation revealed that:

- All 12 flights had information about their second alternate airport and landing time.
- 9 flights did not reach their original destinations, and none had a recorded departure time from the second alternate airport.
- 3 flights eventually reached their original destinations, and all 3 had a recorded departure time from the second alternate airport.

This pattern further indicates that the availability of operational information is associated with the final status of the flight.

## 4. Investigation of Unusual Records

During the analysis, 82 records were identified with alternate airport information even though they were not marked as diverted flights.

Further investigation showed that all 82 records had a special value in the alternate landing count field, indicating an air return to the origin airport. In all cases, the recorded alternate airport was the same as the original departure airport.

Among these records:

- 53 flights were cancelled.
- 29 flights were not cancelled and had recorded arrival information.

The 29 non-cancelled records were unusual because they shared the same airline and had incomplete additional operational information.

These records were classified as potential data inconsistencies requiring further investigation. However, they were not automatically considered errors, modified, or removed.

## 5. Final Decisions

The analysis showed that many missing values in diversion-related information are associated with actual flight operations.

For example, when a flight does not continue its journey to the original destination, some subsequent operational information may not be available.

Therefore:

- Missing values were not automatically replaced with zero.
- Records were not removed simply because some operational information was missing.
- Unusual records were identified for further investigation.
- Missing values were evaluated based on the operational status of each flight.

**Conclusion:** Understanding missing values requires more than checking their frequency. The operational context of each flight must also be considered to avoid incorrectly modifying or removing valid records.

# Data Validation Report

## 1. Introduction

After completing the Missing Value Analysis, a comprehensive Data Validation process was conducted to evaluate the reliability, consistency, and logical correctness of the dataset.

The original dataset contained 608,389 flight records. After removing the completely empty `Unnamed: 109` column, the dataset contained 109 original columns. Two additional derived features, `TotalDelayCauses` and `DelayDifference`, were subsequently created, resulting in a dataset with 111 columns.

The validation process focused on data types, numerical ranges, logical relationships between columns, time consistency, flight performance metrics, and route distance consistency.

The main objective was to identify confirmed data errors while distinguishing them from unusual but potentially valid operational records.

---

## 2. Data Structure and Column Validation

The dataset was divided into eight logical subsets to make the validation process more manageable.

The column allocation was verified to ensure that:

- All 109 original columns were assigned to a subset.
- No original column was left unassigned.
- No column was assigned more than once.
- No duplicate column names were identified in the original dataset.

The eight subsets covered flight time, airline information, delays, performance, cancellations and diversions, delay causes, and additional diversion information.

---

## 3. Data Type Validation

Data types were reviewed and adjusted where necessary.

The following changes were applied:

- `FlightDate` was converted to datetime.
- `Cancelled`, `Diverted`, `DepDel15`, `ArrDel15`, and `DivReachedDest` were converted to nullable Boolean types.
- Airport, airline, flight, and other identifier columns were converted to string types.

HHMM-based time columns were kept numeric because their values represent time in a specific numerical format rather than standard decimal measurements.

Five time columns contained values equal to `2400`:

| Column | Number of Records |
|---|---:|
| DepTime | 82 |
| ArrTime | 306 |
| WheelsOff | 102 |
| WheelsOn | 287 |
| FirstDepTime | 3 |

The value `2400` represents midnight. These values were not modified because changing them requires considering the associated flight date and possible date rollover.

---

## 4. Missing Value and Structural Consistency Validation

Missing values were evaluated across all eight logical subsets.

The investigation showed that many missing values were associated with specific flight statuses rather than random data loss.

### 4.1 Cancellation and Diversion Relationships

- Cancelled flights naturally lacked several actual departure, arrival, and performance measurements.
- Diverted flights often lacked arrival information for the original destination.
- Normal flights generally had no diversion-related information.
- The 461 diverted flights that did not reach their original destination lacked `DivArrDelay` and `DivActualElapsedTime`.

These missing values were considered structurally expected.

### 4.2 Delay Cause Fields

The five delay cause columns contained 448,753 missing values each.

Further investigation showed:

- Flights with `ArrDel15 = 0` had missing delay cause values.
- Flights with `ArrDel15 = 1` had populated delay cause values.
- Flights with missing `ArrDel15` also had missing delay cause values.

These missing values were not automatically replaced with zero because missing information does not necessarily mean that a delay cause had a value of zero.

### 4.3 Diversion Information

Diversion-related columns were examined according to the number of diversion landings and whether the flight reached its original destination.

The following patterns were identified:

- Third, fourth, and fifth diversion fields were completely empty.
- There were 12 flights with two alternate landings.
- Some flights that did not reach their original destination lacked subsequent departure information.
- The value `DivAirportLandings = 9` represented Air Return records, not nine alternate landings.

No missing values were imputed or removed during this investigation.

---

## 5. Delay Data Validation

Several checks were performed to validate the relationships between delay-related columns.

### 5.1 Delay Minutes

The following relationships were verified:

- `DepDelayMinutes = max(DepDelay, 0)`
- `ArrDelayMinutes = max(ArrDelay, 0)`

All complete records satisfied these relationships.

No negative values were found in `DepDelayMinutes` or `ArrDelayMinutes`.

### 5.2 Delay Groups

The departure and arrival delay group columns were checked against their corresponding delay-minute values.

The observed groups ranged from -2 to 12, with no unexpected group values.

The positive group intervals followed the expected structure:

| Group | Delay Interval |
|---:|---|
| 0 | 0–14 minutes |
| 1 | 15–29 minutes |
| 2 | 30–44 minutes |
| 3 | 45–59 minutes |
| ... | ... |
| 11 | 165–179 minutes |
| 12 | 180 minutes or more |

Groups -2 and -1 both corresponded to zero delay minutes, representing different early/on-time statuses that cannot be distinguished using the clipped delay-minute columns alone.

### 5.3 Delay Cause Consistency

The sum of the five delay causes was compared with `ArrDelayMinutes` for delayed flights.

Results:

- 159,607 records matched exactly.
- 29 records had a mismatch.

All 29 mismatches belonged to Southwest Airlines (`WN`). In these records, all five delay cause values were zero despite positive arrival delay minutes.

These records represented approximately 0.085% of Southwest's delayed flights.

They were classified as potential reporting inconsistencies and were not modified.
